```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("messy_sales - Sheet1.csv")
df.head(10)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python

df["Units Sold"] = pd.to_numeric(df["Units Sold"], errors = "coerce")
df["Total Sales"] = pd.to_numeric(df["Total Sales"], errors = "coerce")
```

```python
df["Units Sold"] = df["Units Sold"].fillna(df["Units Sold"].mean()).round(2)
df["Total Sales"] = df["Units Sold"].fillna(df["Total Sales"].mean()).round(2)
```

```python
df.columns = df.columns.str.strip()
df.head(10)
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='mixed')
df.head(10)
```

```python
for col in ["Region","Product","Category"]:
    df[col] = df[col].str.strip().str.title()
```

```python
df.head(10)
```

```python
for col in ["Unit Price","Total Sales"]:
    df[col] = df[col].astype(str)
    df[col] = (df[col].str.replace("[$,USD]","",regex=True).str.strip())
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df["Profit Margin (%)"] = ( df["Profit Margin (%)"].astype(str).str.replace("%","",regex=True).str.replace("percent","",regex = True).astype(float))
df["Profit Margin (%)"] = np.where(df["Profit Margin (%)"]<1,df["Profit Margin (%)"]*100,df["Profit Margin (%)"])
```

```python
df_cleaned = df.dropna(subset=["Date","Total Sales","Region"])
```

```python
df.head(10)
```

```python
df.loc[df['Category'].str.startswith('E'), 'Category'] = "Electronics"

df.loc[df['Category'].str.startswith('G'), 'Category'] = "Gadgets"
```
