```python
import pandas as pd
import json

df = pd.read_csv('smartphones - smartphones.csv')
```

```python
df.columns = df.columns.str.strip().str.lower()

df['price'] = df['price'].str.replace('₹', '').str.replace(',', '').astype(float)

df['rating'] = df['rating'].fillna(df['rating'].mean())
df['os'] = df['os'].fillna('Unknown')
df['card'] = df['card'].fillna('Not Specified')
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
def extract_brand(model_name):
    return str(model_name).split(' ')[0]

df['brand'] = df['model'].apply(extract_brand)
```

```python
def extract_ram(ram_string):
    try:
        parts = str(ram_string).split('GB')
        return int(parts[0].strip())
    except:
        return 0

df['ram_gb'] = df['ram'].apply(extract_ram)
```

```python
def extract_battery(battery_string):
    try:
        parts = str(battery_string).split('mAh')
        return int(parts[0].strip())
    except:
        return 0

df['battery_mah'] = df['battery'].apply(extract_battery)
```

```python
def categorize_price(price):
    if price < 15000:
        return 'Budget'
    elif price <= 40000:
        return 'Mid-Range'
    else:
        return 'Premium'

df['price_segment'] = df['price'].apply(categorize_price)
```

```python
df[['model', 'brand', 'ram_gb', 'battery_mah', 'price_segment']].head()
```

```python
df.head()
```

```python
total_price = df['price'].sum()
avg_rating = df['rating'].mean()
unique_brands = df['brand'].nunique()
```

```python
avg_price_brand = df.groupby('brand')['price'].mean().sort_values(ascending=False).head(10).reset_index()
```

```python
brand_count = df.groupby('brand')['model'].count().sort_values(ascending=False).reset_index()
```

```python
top_rated_phones = df.sort_values(by=['rating', 'price'], ascending=[False, True]).head(10)[['model', 'price', 'rating']]
```

```python
avg_battery_segment = df.groupby('price_segment')['battery_mah'].mean().reset_index()
```

```python
kpi_summary = {
    "TOTAL_PRICE": float(total_price),
    "AVERAGE_RATING": float(avg_rating),
    "TOTAL_UNIQUE_BRANDS": int(unique_brands)
}
```

```python
with open("smartphone_kpi_summary.json", "w") as f:
    json.dump(kpi_summary, f, indent=4)
```

```python
avg_price_brand.to_json("avg_price_by_brand.json", orient="records", indent=4)
brand_count.to_json("models_per_brand.json", orient="records", indent=4)
top_rated_phones.to_json("top_rated_smartphones.json", orient="records", indent=4)
```
